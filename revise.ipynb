```python
import pandas as pd

data = [
    [56, 70, 90, 40, 35],
    [67, 0, 92, 52, 50],
    [45, 82, 95, 48, 55]
]

students = ["Rahul", "Aisha", "Karan", "Halima", "Lisa"]
subjects = ["sub1", "sub2", "sub3"]

df = pd.DataFrame(data=data, index=subjects, columns=students)
print(df)

for i in range(5):  # loop through each student
    skip_student = False

    # --- check if any mark is 0 for this student ---
    for j in range(3):
        if df.iloc[j, i] == 0:
            skip_student = True
            break

    if skip_student:
        print(students[i], "skipped due to zero marks")
        continue


    Sum = 0
    for j in range(3):
        Sum += df.iloc[j, i]

    avg = Sum / 3

    if avg >=90:
        print("Average above 90. Program stopped.")
        break
    
    # --- print pass/fail ---
    if avg > 50:
        print(students[i], "Pass")
    elif avg <50:
        print(students[i], "Failed")
    print("Student:", students[i], "Average:", avg)
```

```python
marks = [
    [56, 70, 90, 40, 35],   # sub1
    [67, 0, 92, 52, 50],    # sub2
    [45, 82, 95, 48, 55]    # sub3
]

students = ["Rahul", "Aisha", "Karan", "Halima", "Lisa"]
subjects = ["sub1", "sub2", "sub3"]

# Loop through each student
for i in range(len(students)):
    skip_student = False

    # --- check if any mark is 0 ---
    for j in range(len(subjects)):
        if marks[j][i] == 0:
            skip_student = True
            break

    if skip_student:
        print(students[i], "skipped due to zero marks")
        continue

    # --- calculate sum ---
    total = 0
    for j in range(len(subjects)):
        total += marks[j][i]

    avg = total / len(subjects)

    # --- stop condition ---
    if avg >= 90:
        print("Average above 90. Program stopped.")
        break

    # --- pass / fail ---
    if avg > 50:
        print(students[i], "Pass")
    else:
        print(students[i], "Failed")

    print("Student:", students[i], "Average:", avg)
```

```python
# Step 1: Append new records (a+ mode)
def add_student():
    f = open("students.txt", "a+")
    name = input("Enter name: ")
    marks = int(input("Enter marks: "))
    f.write(f"{name},{marks}\n")  # Format: Name,Marks
    f.close()

# Step 2: Read & filter (r mode)
def display_toppers():
    f = open("students.txt", "r")
    for line in f.readlines():
        name, marks = line.strip().split(',')
        if int(marks) >=75:
            print(f"{name} - {marks}")
    f.close()

# Menu
while True:
    print("\n1. Add Student\n2. Display Toppers\n3. Exit")
    choice = input("Choose: ")
    if choice == "1":
        add_student()
    elif choice == "2":
        display_toppers()
    else:
        break
```

```python
def word_stats(text):
    vowels = "aeiouAEIOU"
    count1 = 0   # vowels
    count2 = 0   # consonants
    count3 = len(text.split())  # words

    for a in text:
        if a.isalpha():
            if a in vowels:
                count1 += 1
            else:
                count2 += 1

    return count1, count2, count3


text = input("Enter a text to find out something: ")

vowel, consonant, word = word_stats(text)

print("no. vowels in text", vowel)
print("no. consonants in text", consonant)
print("no. words in text", word)
```
